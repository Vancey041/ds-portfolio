$$x = \begin{bmatrix} 1 \\ 0 \\ 1 \end{bmatrix}$$

$$y = \begin{bmatrix} 1 \end{bmatrix}$$

$$f=max(0, Z_n)$$

$$\text{hidden unit weights} = 
\begin{bmatrix}
w_{11} = 0.2 && w_{12} = -0.3 \\
w_{13} = 0.4 && w_{14} = 0.1 \\
w_{15} = -0.5 && w_{16} = 0.2
\end{bmatrix}$$

$$\text{output unit weights} = 
\begin{bmatrix}
w_{21} = -0.3 \\
w_{22} = -0.2
\end{bmatrix}$$

$$\theta = 
\begin{bmatrix}
\theta_{1} = -0.4 \\
\theta_{2} = 0.2 \\
\theta_{3} = 0.1
\end{bmatrix}$$
</div>

```python
import numpy as np

# Inputs and Targets
x = np.array([1, 0, 1])
y = np.array([1])

# Weights and Biases
w_hidden = np.array([[ 0.2, -0.3],
                     [ 0.4,  0.1],
                     [-0.5,  0.2]])

w_output = np.array([-0.3, -0.2])

theta_hidden = np.array([-0.4, 0.2])
theta_output = np.array([0.1])

# Hidden Layer Forward Pass
Z_hidden = np.dot(x, w_hidden) + theta_hidden
H = np.maximum(0, Z_hidden)  # ReLU activation

# Output Layer Forward Pass
Z_output = np.dot(H, w_output) + theta_output
y_hat = np.maximum(0, Z_output)  # ReLU activation

# Error Calculation
delta = y_hat - y

print(f"Hidden Layer Activations (H): {H}")
print(f"Predicted Output (y_hat): {y_hat}")
print(f"Error (delta): {delta}")
```
